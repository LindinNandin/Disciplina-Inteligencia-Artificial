##Exercicio 19
###19.	Crie dois arrays de nu´meros inteiros aleat´orios de tamanho 20 e calcule o ˆangulo entre eles (produto escalar normalizado).



```python
import numpy as np

array1 = np.random.randint(1, 50, size=20)
array2 = np.random.randint(1, 50, size=20)

produtoescalar = np.dot(array1, array2)

norma1 = np.linalg.norm(array1)
norma2 = np.linalg.norm(array2)

cos = produtoescalar / (norma1 * norma2)

angulorad = np.arccos(cos)
angulograus = np.degrees(angulorad)

print(f"Produto escalar: {produtoescalar}")
print(f"Cosseno do ângulo: { cos}")
print(f"Ângulo (radianos): { angulorad}")
print(f"Ângulo (graus) {angulograus}")
```

###16. Resolva a equacao diferencial de segunda ordem d2y +y = 0, com condicoes iniciais y(0) = 1 e y′(0) = 0.

```python
import numpy as np
from scipy.integrate import solve_ivp

# EDO de 2ª ordem: y'' + y = 0
# Convertendo para um sistema :
#   y1 = y  -> y1' = y2
#   y2 = y' ->y2' = -y1
def sistema(x, estado):
    y1, y2 = estado
    dy1dx = y2
    dy2dx = -y1
    return [dy1dx, dy2dx]

estado_inicial = [1, 0]

# Resolver a EDO no intervalo [0, 10]
x_avaliar = np.linspace(0, 10, 1000)
solucao = solve_ivp(sistema, (0, 10), estado_inicial, t_eval=x_avaliar, rtol=1e-8)

y = solucao.y[0]  # y1 corresponde a y(x)

print("y(10) =", y[-1])
```

##Exercicio 19
###19. Realize a transformada de Fourier de um sinal discreto composto de senos e cossenos.

```python
import numpy as np
from scipy.fft import fft, fftfreq


fs = 50            # taxa de amostragem
N = 100            # amostras
n = np.arange(N)   # índices discretos
t = n / fs         # tempo pra cada amostra

# Sinal discreto composto por senos e cossenos de diferentes frequências
sinal = (3*np.sin(2*np.pi*5*t)      # seno, 5 Hz, amplitude 3
       + 2*np.cos(2*np.pi*10*t)     # cosseno, 10 Hz, amplitude 2
       + 1*np.sin(2*np.pi*15*t))    # seno, 15 Hz, amplitude 1

# Calcular a Transformada Discreta de Fourier
X = fft(sinal)
freqs = fftfreq(N, 1/fs)

metade = N // 2
freqs_pos = freqs[:metade]
magnitude = np.abs(X[:metade]) / N * 2

print("Frequências dominantes:", freqs_pos[magnitude > 0.1])
```

##Exercicio 20
###20. Resolva um sistema de equacoes lineares com tres variaveis utilizando SciPy.

```python
import numpy as np
from scipy.linalg import solve

# Sistema é o seguinte:
#   2x + 3y - z  = 5
#   4x - y  + 2z = 6
#  -2x + y  + 2z = 3

#separa os coeficientes
A = np.array([
    [2,  3, -1],
    [4, -1,  2],
    [-2, 1,  2]
])

#separa os termos independentes
b = np.array([5, 6, 3])

# resolve o sistema
x, y, z = solve(A, b)

print(f"x = {x}")
print(f"y = {y}")
print(f"z = {z}")
```

##Exercicio 8
###8. Plote um grafico de linha com intervalos de confianca (sombramento) para a funcao y = sin(x).

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 500)
y = np.sin(x)

# Simula uma margem de incerteza
margemerro = 0.15 + 0.1 * np.abs(np.sin(x / 2))

limitesuperior = y + margemerro
limiteinferior = y - margemerro

plt.figure(figsize=(9, 6))

plt.plot(x, y, color='navy', linewidth=2, label='y = sin(x)')

plt.fill_between(x, limiteinferior, limitesuperior,
                  color='Pink', label='Intervalo de confiança')

plt.title('Gráfico de y = sin(x) com Intervalo de Confiança')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, linestyle='--', alpha=0.4)
plt.legend()

plt.show()
```
