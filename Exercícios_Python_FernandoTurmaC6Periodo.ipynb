```python
#21. Busca Binária

def binaria_busca(lista, procurando):
    inicio = 0
    fim = len(lista) - 1

    #esse algoritmo busca de metade a lista em metade, comparando o numero com o meio da lista, se for maior ele pega a metade direita, se for menor ele pega a metade esquerda, e assim por diante até achar o numero ou acabar a lista
    while inicio <= fim:

        #procura o mein da lista
        meio = (inicio + fim) // 2

        #compara o numero do meio com o numero que esta procurando e confere se é maior ou menor
        if lista[meio] == procurando:
            return meio
        elif lista[meio] < procurando:
            inicio = meio + 1
        else:
            fim = meio - 1

    #retorna erro caso o numero nao seja encontrado na lista
    return "erro"


lista = [ 1, 25, 33, 78, 99, 105, 127]
procurando = 78

resultado = binaria_busca(lista, procurando)

if resultado != "erro":
    print(f"Elemento {procurando} encontrado no índice {resultado}.")
else:
    print("Elemento não encontrado na lista.")
```

```python
#22. Criptografia de Cifra de César

alfabeto_minusculo = "abcdefghijklmnopqrstuvwxyz"
alfabeto_maiusculo = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

mensagem = input("Digite a mensagem: ")
deslocamento = int(input("Digite sua chave : "))

codificada = ""

#para cada letra na mensagem, testa se é maiuscula ou minuscula, e move ela para frente no alfabeto, usa o .find pra achar a posicao da letra no alfabeto, usa o % 26 pra dar loop e voltar do z para o a
for letra in mensagem:

    if letra in alfabeto_minusculo:
        posicao = alfabeto_minusculo.find(letra)
        nova_posicao = (posicao + deslocamento) % 26
        codificada += alfabeto_minusculo[nova_posicao]

    elif letra in alfabeto_maiusculo:
        posicao = alfabeto_maiusculo.find(letra)
        nova_posicao = (posicao + deslocamento) % 26
        codificada += alfabeto_maiusculo[nova_posicao]

    #se for espaco ele nao muda nada
    else:
        codificada += letra

print(f"Mensagem codificada: {codificada}")
```

```python
#30. jogo da forca

#no tutorial que eu vi tinha boneco mas nao sou artista ent fica assim
def mostrar_forca(erros):
    etapas = [
        """
        Erros: 0/6 - sem boneco
        """,
        """
        Erros: 1/6 - cabeça do boneco
        """,
        """
        Erros: 2/6 - tronco do boneco
        """,
        """
        Erros: 3/6 - braço do boneco
        """,
        """
        Erros: 4/6 - outro braço do boneco
        """,
        """
        Erros: 5/6 - perna do boneco
        """,
        """
        Erros: 6/6 - outra perna do boneco
        """
    ]
    return etapas[erros]

def forca():
    palavra = "escoliose"
    letras_descobertas = ["_","_","_","_","_","_","_","_","_"]
    letras_tentadas = []
    erros = 0
    max_erros = 6

    print("=== JOGO DA FORCA ===")

    #enquanto o usuario nao errou 6 vezes e ainda tem "_" na lista de letras descobertas, ele continua pedindo para o usuario digitar uma nova letra
    while erros < max_erros and "_" in letras_descobertas:
        print(mostrar_forca(erros))
        print("Palavra: " + " ".join(letras_descobertas))
        print(f"Letras tentadas: {', '.join(letras_tentadas)}")
        print(f"Erros: {erros}/{max_erros}")

        letra = input("Digite uma letra: ").lower()

        #garante que o usuario nao va digitar uma virgula ou algo assim
        if len(letra) != 1 or not letra.isalpha():
            print("Digite apenas letras\n")
            continue

        #garante que o usuario nao va perder a tentativa por digitar a mesma letra mais de uma vez
        if letra in letras_tentadas:
            print("Essa letra ja foi!\n")
            continue

        #coloca a letra digitada na lista de letras tentadas
        letras_tentadas.append(letra)

        #verifica se acertou uma letra e caso tenha coloca na lista de "___" na posicao que se alinha, completando no visual
        #caso usuario erre, ele soma 1 ao contador de erros
        if letra in palavra:
            for i, l in enumerate(palavra):
                if l == letra:
                    letras_descobertas[i] = letra
            print("Letra encontrada.\n")
        else:
            erros = erros + 1
            print("A palavra nao tem essa letra.\n")

    print(mostrar_forca(erros))

    #confere se todos os "_" foram substituidos por letras, caso sim, o usuario ganhou, caso nao, ele perdeu
    if "_" not in letras_descobertas:
        print(f"Parabéns! Voce ganhou!")
    else:
        print(f"Você perdeu! O boneco morreu :")

forca()
```

```python
#31. Ordenação por Bolha

organizar = [73, 18, 42, 7, 56, 31, 89, 14]

#Passa pelo array, um valor por vez, para cada valor, compara o com o valor seguinte, se o valor for maior, troca os valores de lugar, Repete ate que todos os valores estejam em ordem crescente.

n = len(organizar)
for i in range(n-1):
  for j in range(n-i-1):
    if organizar[j] > organizar[j+1]:
      organizar[j], organizar[j+1] = organizar[j+1], organizar[j]

print(organizar)
```

```python
##32 JOGO DA VELHA

#cria uma lista com 9 elementos para representar cada espaco do jogo da velha, usa o " " para representar um slot nao atribuido do tabuleiro
def criartabuleiro():
    return [" " for _ in range(9)]

#Faz o visual em 3x3 do jogo da velha e caso usa os itens do tabuleiro para mostrar o tabuleiro sendo preenchido
def mostrartabuleiro(tabuleiro):
    print()
    print(f" {tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]} ")
    print("---+---+---")
    print(f" {tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]} ")
    print("---+---+---")
    print(f" {tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]} ")
    print()

#funcao para checar a vitoria do jogador
def checarvencedor(tabuleiro, jogador):

    #essas sao todas as combinacoes de 3 casas que formam uma linha, considerando as cordenadas da lista/tabuleiro
    combinacoesvencedoras = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8], [0, 3, 6], [1, 4, 7], [2, 5, 8], [0, 4, 8], [2, 4, 6]
    ]

    #ele testa cada uma das combinacoes acima e confere se todas elas sao iguais ao simbolo do jogador, caso sejam, retorna que ele ganhou
    for combinacao in combinacoesvencedoras:
        if all(tabuleiro[i] == jogador for i in combinacao):
            return True
    return False

#checa se o tabuleiro ficou cheiro e o jogo resultou em empate
def tabuleiro_cheio(tabuleiro):
    return " " not in tabuleiro

#funcao do jogo em si
def jogarjogo():

    #cria o tabuleiro vazio e define que o X começa
    tabuleiro = criartabuleiro()
    jogador_atual = "X"

    print("=== JOGO DA VELHA ===")
    print("Posições do tabuleiro:")
    print(" 1 | 2 | 3 ")
    print("---+---+---")
    print(" 4 | 5 | 6 ")
    print("---+---+---")
    print(" 7 | 8 | 9 ")

    while True:
        mostrartabuleiro(tabuleiro)
        print(f"Vez do jogador: {jogador_atual}")

        #tem que tirar 1 do numero digitado pq os arrays começam no zero, usa try para nao dar erro caso o usuario digite alguma coisa
        try:
            posicao = int(input("Escolha uma posição (1-9): ")) - 1
        except ValueError:
            print("Digite um número válido!\n")
            continue

        #verifica se o numero esta em uma casa existente
        if posicao < 0 or posicao > 8:
            print("Posição inválida! Escolha entre 1 e 9.\n")
            continue

        #verifica se a casa esta vazia
        if tabuleiro[posicao] != " ":
            print("Essa posição já está ocupada! Escolha outra.\n")
            continue

        #coloca na lista de tabuleiro o simbolo do jogador (X ou O),
        tabuleiro[posicao] = jogador_atual

        #chama a funcao de checar vitoria e avisa caso o jogador tenha ganhado
        if checarvencedor(tabuleiro, jogador_atual):
            mostrartabuleiro(tabuleiro)
            print(f"O {jogador_atual} venceu!")
            break

        #confere se o tabuleiro esta cheio para chegar o empate
        if tabuleiro_cheio(tabuleiro):
            mostrartabuleiro(tabuleiro)
            print("Deu velha!")
            break

        jogador_atual = "O" if jogador_atual == "X" else "X"

jogarjogo()
```

```python
#Funções Recursivas

#funcao recursiva sao funcoes que chamam elas mesmas

#essa funcao chama ela mas diminuindo o valor de forma que gera a parte do fatorial até chegar no um, depois ele multiplica e retorna a resposta do fatorial
def fatorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * fatorial(n - 1)

numero = int(input("Digite um número: "))

if numero < 0:
    print("Fatorial não é definido para números negativos.")
else:
    resultado = fatorial(numero)
    print(f"O fatorial é {resultado}")
```

```python
#40. Lista de Compras

lista_compras = []

#funcao para adicionar item na sua lista
def adicionar_item():
    item = input("Digite o nome do item: ")
    lista_compras.append(item)
    print(f'"{item}" adicionado à lista!')

#funcao para remover os items
def remover_item():

    #mostra os itens
    listar_itens()

    try:
        indice = int(input("Digite o número do item a remover: ")) - 1
        #se encontrar o item remove ele
        if 0 <= indice < len(lista_compras):
            item_removido = lista_compras.pop(indice)
            print(f'"{item_removido}" removido da lista!')
        else:
            print("Número inválido!")
    except ValueError:
        print("Digite um número válido!")

def listar_itens():

    #confere se a lista nao tem nada e avisa na tela
    if not lista_compras:
        print("A lista de compras está vazia.")

    #caso a lista tenha algo nela mostra todos os items e seus numeros
    else:
        print("\n--- Itens na lista ---")
        for i, item in enumerate(lista_compras, start=1):
            print(f"{i}. {item}")

#faz com que sempre volte pro menu
while True:

    print("\n=== LISTA DE COMPRAS ===")
    print("[1] - Adicionar item")
    print("[2] - Remover item")
    print("[3] - Listar itens")
    print("[4] - Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        adicionar_item()
    elif opcao == "2":
        remover_item()
    elif opcao == "3":
        listar_itens()
    elif opcao == "4":
        print("Saindo")
        break
    else:
        print("Opção inválida!")
```

```python
#42. Números Primos até N

#funcao pra verificar se o numero e primo
def verificarprimo(numero):
    if numero < 2:
        return False
    for contador in range(2, numero):
        if numero % contador == 0:
            return False
    return True

limite = int(input("Digite ate que numero deseja verificar: "))

primos = []

for numero in range(2, limite + 1):
    if verificarprimo(numero):
        primos.append(numero)

print("Números primos até", limite, ":", primos)
```

```python
#47. Contar Números em Lista

numeros = [21, 33, 21, 21, 9, 19, 9, 19, 7, 2, 99]

contagem = {}

#passa por cada numero na lista ai o "contagem.get(numero, 0)" verifica se o numero atual ja esta no dicionário, se ele ainda não existir, retorna 0 como padrão e adiciona 1, entao se o numero ja existir no dicionário, ele pega o valor atual e adiciona 1 a ele
for numero in numeros:
    contagem[numero] = contagem.get(numero, 0) + 1

print(f"Lista original: {numeros}")
print("Contagem de cada número:")

#imprime os numeros e suas respectivas quantidades de vezes que aparecem na lista
for numero, quantidade in contagem.items():
    print(f"  {numero}: {quantidade} vezes")
```

```python
#48. Ordenação por Inserção

#Insertion sort funciona da seguinte forma: ele percorre a lista do segundo elemento até o último, e para cada elemento, ele compara com os elementos anteriores e insere na posição correta, assim "inserindo" cada item no local correto

lista = [42, 7, 91, 15, 63, 28, 84, 3, 56, 77]

tamanho = len(lista)


for item in range(1,tamanho):

    indexinsercao = item

    # remove o elemento da  posição i e guarda o valor
    valoratual = lista.pop(item)

    # percorre de trás pra frente, do i-1 até 0
    for j in range(item-1, -1, -1):
        if lista[j] > valoratual:
            #ao encontrar um elemento menor que o valoratual, ele define a posição de inserção como a posição do elemento menor encontrado
            indexinsercao = j

    #insere o valoratual na posição de inserção encontrada
    lista.insert(indexinsercao, valoratual)

print(lista)
```

```python
#49. Soma dos Dígitos

numero = int(input("Digite um número inteiro: "))

soma = 0
temp = numero

while temp > 0:
    #pega o ultimo numero do numero escrito, usando o resto da divisao por 10
    digito = temp % 10

    #adiciona na soma total o digito que foi removido
    soma += digito

    #remove o ultimo digito, por meio de uma divisao inteira, entao o resto é descartado
    temp //= 10

print(f"A soma dos dígitos de {numero} é {soma}")
```

```python
#50. Cálculo de Frequência de Letras

texto = input("Digite uma frase: ").lower()

frequencia = {}

for letra in texto:

    #isalpha confere se realmente é uma letra
    if letra.isalpha():
        #passa por cada letra na lista ai o "frequencia.get(letra, 0)" verifica se a letra atual ja esta no dicionário, se ele ainda não existir, retorna 0 como padrão e adiciona 1, entao se a letra ja existir no dicionário, ele pega o valor atual e adiciona 1 a ele
        frequencia[letra] = frequencia.get(letra, 0) + 1

#total para calcular a frequencia em porcentagem
total_letras = sum(frequencia.values())

print(f"Frase digitada: {texto}")
print("====================")
print("Frequência de cada letra:")

#passa por cada letra no dicionario ordenando pela frequencia absoluta dos itens
for letra, quantidade in sorted(frequencia.items()):
    frequenciaporcentagem = (quantidade / total_letras) * 100
    print(f"  {letra}: {quantidade} ({frequenciaporcentagem:.2f}%)")
```
